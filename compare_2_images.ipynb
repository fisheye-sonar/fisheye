```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os

dir_1 = "/home/mahobley/Code/fisheye/"
dir_2 = "/home/mahobley/Code/fisheye_main/"
# dir_2 = "/home/mahobley/Code/fisheye/no_resize/"

fns = os.listdir(dir_1)
fn_imgs = [fn for fn in fns if fn.endswith(".png") and fn.startswith("img_")]

for i, fn in enumerate(fn_imgs):
    img1 = cv2.imread(dir_1 + fn)
    img2 = cv2.imread(dir_2 + fn)

    # as numpy arrays
    img1 = np.array(img1).astype(np.float32)/255.0
    img2 = np.array(img2).astype(np.float32)/255.0
    print(f"{img1.shape=} {img2.shape=}")
    diff = img1 - img2
    abs_diff = np.abs(diff)
    diff_norm = (diff + 1)/2
    print(f"{np.sum(diff)} {np.min(diff)} {np.max(diff)}")
    if i  == 0:
        fig, axs = plt.subplots(1, 2, figsize=(20, 20))
        axs[0].imshow(img1[600:800,150:350,0])
        axs[0].set_title("img1")
        axs[1].imshow(img2[600:800, 150:350, 0])
        axs[1].set_title("img2")
        plt.show()
        fig, axs = plt.subplots(1, 2, figsize=(20, 20))

        axs[0].imshow((diff_norm[600:800, 150:350, 0]))
        axs[0].set_title("img1 - img2")
        axs[1].imshow(abs_diff[600:800, 150:350, 0])
        axs[1].set_title("np.abs(img1 - img2)")
        plt.show()
```
